```python

import pandas as pd

orders = pd.DataFrame({
    'Order_ID': [101, 102, 103, 104, 105],
    'Customer_ID': [1, 2, 1, 3, 4],
    'Sales': [500, 800, 300, 900, 400]
})

customers = pd.DataFrame({
    'Customer_ID': [1, 2, 3, 5],
    'Customer_Name': ['Ravi', 'Priya', 'Amit', 'Sneha'],
    'City': ['Hyderabad', 'Bengaluru', 'Chennai', 'Mumbai']
})
```

```python
#Merge orders and customers to display customer names and cities alongside each order. Keep only matching customer IDs.
df = pd.merge(orders, customers, on='Customer_ID', how='inner')
df[['Customer_Name','City']]
```

```python
#Perform a left merge so that every order remains in the output, even if its customer details don't exist.
```

```python
df = pd.merge(orders, customers, on='Customer_ID', how='left')
df[['Order_ID', 'Customer_ID', 'Sales', 'Customer_Name', 'City']]
```

```python
#Identify the order with a missing customer name.
df = pd.merge(
    orders, customers,
    on='Customer_ID',
    how='left'
)

missing_customers = df[df['Customer_Name'].isna()]

print(missing_customers)
```
